````python
# Extract Date-Time Features
df['Year'] = df['Timestamp'].dt.year
df['Month'] = df['Timestamp'].dt.month
df['Day'] = df['Timestamp'].dt.day
df['Hour'] = df['Timestamp'].dt.hour

df[['Year', 'Month', 'Day', 'Hour']].head()
````

````python
# Label Encoding
from sklearn.preprocessing import LabelEncoder

# Separate encoders are kept (instead of one reused encoder) so that both
# can be saved and reused later for deployment / new incoming data.
pump_status_encoder = LabelEncoder()
fault_code_encoder = LabelEncoder()

df['PumpStatus'] = pump_status_encoder.fit_transform(df['PumpStatus'])
df['FaultCode'] = fault_code_encoder.fit_transform(df['FaultCode'])

df[['PumpStatus', 'FaultCode']].head()
````

###  Observation

Categorical features have been converted into numerical values using Label Encoding, making them compatible with Deep Learning algorithms.

````python
# Drop Timestamp Column
df.drop('Timestamp', axis=1, inplace=True)
df.head()
````

````python
# Feature Scaling
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns
df[numerical_columns] = scaler.fit_transform(df[numerical_columns])
df.head()
````

Feature scaling standardizes numerical variables so that they have a mean of approximately zero and a standard deviation of one. This improves the convergence and performance of Deep Learning models.

# 13. Train-Test Split

The train-test split is an essential step in the Deep Learning workflow. It divides the dataset into two subsets:

- **Training Set:** Used to train the Deep Learning models.

#  16. One-Dimensional Convolutional Neural Network (1D CNN)

A One-Dimensional Convolutional Neural Network (1D CNN) is a Deep Learning architecture capable of identifying local patterns and relationships within sequential or structured numerical data.

In this project, a 1D CNN is developed to predict the water output of the smart wind-powered water pumping system. The model uses environmental, electrical, mechanical, and engineered sensor features as input.

The performance of the 1D CNN model will be evaluated using Mean Absolute Error (MAE), Mean Squared Error (MSE), Root Mean Squared Error (RMSE), and R² Score. The results will later be compared with the ANN and RNN models.

````python
# Prepare Data for 1D CNN
X_train_cnn = X_train_reg.values.reshape(X_train_reg.shape[0],X_train_reg.shape[1],1)

X_test_cnn = X_test_reg.values.reshape(X_test_reg.shape[0],X_test_reg.shape[1],1)

print("X_train_cnn Shape:", X_train_cnn.shape)
print("X_test_cnn Shape :", X_test_cnn.shape)
````

````python
# Build 1D CNN Model
cnn_model = Sequential()

cnn_model.add(Conv1D(filters=64,kernel_size=3,activation='relu',input_shape=(X_train_cnn.shape[1], 1)))

cnn_model.add(MaxPooling1D(pool_size=2))

cnn_model.add(Conv1D(filters=32,kernel_size=3,activation='relu'))

cnn_model.add(Flatten())

cnn_model.add(Dense(32, activation='relu'))

cnn_model.add(Dropout(0.2))

# Output Layer
cnn_model.add(Dense(1))

cnn_model.summary()
````

````python
# Compile 1D CNN Model
cnn_model.compile(optimizer='adam',loss='mse',metrics=['mae'])
````

````python
# Train 1D CNN Model
history_cnn = cnn_model.fit(X_train_cnn,y_train_reg,validation_split=0.2,epochs=50,batch_size=32,verbose=1)
````

````python
plt.figure(figsize=(10,5))
plt.plot(history_cnn.history['loss'],label='Training Loss')
plt.plot(history_cnn.history['val_loss'],label='Validation Loss')
plt.title('1D CNN Training and Validation Loss',fontsize=16)
plt.xlabel('Epoch')
plt.ylabel('Loss')
plt.legend()
plt.tight_layout()
plt.savefig("Graphs/22_1D_CNN_Training_and_Validation_Loss.png",dpi=300,bbox_inches="tight")
plt.show()
````

Six deep learning models were trained in this project: three architectures (ANN, RNN, 1D CNN) each for water-output regression and for fault-code classification. This section packages the **ANN water-output model** (the model used in the Final Prediction section) and the **best-performing fault-detection model** into a Streamlit web application, together with the saved `StandardScaler` and the two `LabelEncoder`s used for `PumpStatus` and `FaultCode`.

**Important limitation — please read before deploying:** the engineered feature `WaterProductionEfficiency` (Section 12) is computed as `WaterOutput_LPH / (PumpPower_W + 1)`, i.e. directly from the regression target, and it is kept as an input feature for *both* the regression and classification models. This is target leakage: in a genuine real-time deployment, `WaterOutput_LPH` for the current reading is exactly what the regression model is being asked to predict, so it cannot also be a known input. The app below works around this by asking the operator for an **assumed / last-known water output** to approximate this feature, and clearly labels the resulting prediction as an approximation. For a production system, retrain the models after removing `WaterProductionEfficiency` (or replace it with a lagged/previous-timestep value) so the leakage is eliminated.

````python
%%writefile app.py
import streamlit as st
import pandas as pd
import numpy as np
import joblib
from datetime import datetime
from tensorflow.keras.models import load_model

st.set_page_config(page_title="Smart Wind-Powered Water Pumping System", page_icon="💨", layout="wide")

# The 17 columns the StandardScaler was originally fit on, in order.
SCALED_COLUMNS = [
    "WindSpeed_mps", "WindDirection_deg", "AirTemperature_C", "Humidity_pct",
    "AirPressure_hPa", "GeneratorRPM", "GeneratorVoltage_V", "GeneratorCurrent_A",
    "BatteryVoltage_V", "PumpStatus", "PumpPower_W", "WaterFlow_LPM",
    "TankLevel_pct", "EnergyGenerated_Wh", "EnergyConsumed_Wh",
    "FaultCode", "WaterOutput_LPH",
]

# Final 24-feature order expected by the regression / classification models.
MODEL_FEATURES = [
    "WindSpeed_mps", "WindDirection_deg", "AirTemperature_C", "Humidity_pct",
    "AirPressure_hPa", "GeneratorRPM", "GeneratorVoltage_V", "GeneratorCurrent_A",
    "BatteryVoltage_V", "PumpStatus", "PumpPower_W", "WaterFlow_LPM",
    "TankLevel_pct", "EnergyGenerated_Wh", "EnergyConsumed_Wh",
    "Year", "Month", "Day", "Hour",
    "WindPowerEfficiency", "PumpEfficiency", "EnergyEfficiency",
    "GeneratorLoadRatio", "WaterProductionEfficiency",
]

FAULT_MODEL_FILES = {
    "ANN": "ANN_FaultDetection_Model.keras",
    "RNN": "RNN_FaultDetection_Model.keras",
    "1D CNN": "CNN_FaultDetection_Model.keras",
}


@st.cache_resource
def load_artifacts():
    scaler = joblib.load("WindPump_StandardScaler.pkl")
    pump_status_encoder = joblib.load("PumpStatus_LabelEncoder.pkl")
    fault_code_encoder = joblib.load("FaultCode_LabelEncoder.pkl")
    ann_reg_model = load_model("ANN_WaterOutput_Model.keras")
    return scaler, pump_status_encoder, fault_code_encoder, ann_reg_model


@st.cache_resource
def load_fault_model(name):
    return load_model(FAULT_MODEL_FILES[name])


scaler, pump_status_encoder, fault_code_encoder, ann_reg_model = load_artifacts()

st.title("💨 Smart Wind-Powered Water Pumping System")
st.caption("Deep learning water-output prediction and fault detection from live sensor readings.")

st.info(
    "⚠️ The trained models rely on an engineered feature (`WaterProductionEfficiency`) "
    "that is derived from water output itself. This app asks for an assumed/last-known water "
    "output value to approximate that feature — treat predictions as indicative, not exact. "
    "See the notebook's Deployment section for details.",
    icon="⚠️",
)

with st.form("sensor_form"):
    st.subheader("Current Sensor Readings")
    c1, c2, c3 = st.columns(3)
    with c1:
        wind_speed = st.number_input("Wind Speed (m/s)", value=5.0)
        wind_dir = st.number_input("Wind Direction (deg)", value=180.0)
        air_temp = st.number_input("Air Temperature (°C)", value=25.0)
        humidity = st.number_input("Humidity (%)", value=50.0)
        air_pressure = st.number_input("Air Pressure (hPa)", value=1013.0)
        gen_rpm = st.number_input("Generator RPM", value=500.0)
    with c2:
        gen_voltage = st.number_input("Generator Voltage (V)", value=220.0)
        gen_current = st.number_input("Generator Current (A)", value=5.0)
        battery_voltage = st.number_input("Battery Voltage (V)", value=12.0)
        pump_status = st.selectbox("Pump Status", options=list(pump_status_encoder.classes_))
        pump_power = st.number_input("Pump Power (W)", value=150.0)
        water_flow = st.number_input("Water Flow (LPM)", value=10.0)
    with c3:
        tank_level = st.number_input("Tank Level (%)", value=60.0)
        energy_gen = st.number_input("Energy Generated (Wh)", value=100.0)
        energy_cons = st.number_input("Energy Consumed (Wh)", value=90.0)
        assumed_water_output = st.number_input(
            "Assumed / Last-Known Water Output (LPH)",
            value=50.0,
            help="Used only to approximate the leaked WaterProductionEfficiency feature. See warning above.",
        )
        timestamp = st.text_input("Timestamp (YYYY-MM-DD HH:MM:SS)", value=datetime.now().strftime("%Y-%m-%d %H:%M:%S"))
        fault_model_choice = st.selectbox("Fault Detection Model", options=list(FAULT_MODEL_FILES.keys()))

    submitted = st.form_submit_button("Run Prediction", type="primary")

if submitted:
    ts = pd.to_datetime(timestamp)
    pump_status_encoded = pump_status_encoder.transform([pump_status])[0]

    # Build the 17-column vector in the exact order the scaler was fit on.
    # FaultCode is unknown at inference time, so a neutral placeholder (0) is used;
    # it is discarded after scaling since it is not one of the 24 model features.
    raw_17 = pd.DataFrame([{
        "WindSpeed_mps": wind_speed, "WindDirection_deg": wind_dir, "AirTemperature_C": air_temp,
        "Humidity_pct": humidity, "AirPressure_hPa": air_pressure, "GeneratorRPM": gen_rpm,
        "GeneratorVoltage_V": gen_voltage, "GeneratorCurrent_A": gen_current,
        "BatteryVoltage_V": battery_voltage, "PumpStatus": pump_status_encoded,
        "PumpPower_W": pump_power, "WaterFlow_LPM": water_flow, "TankLevel_pct": tank_level,
        "EnergyGenerated_Wh": energy_gen, "EnergyConsumed_Wh": energy_cons,
        "FaultCode": 0, "WaterOutput_LPH": assumed_water_output,
    }])[SCALED_COLUMNS]

    scaled_17 = scaler.transform(raw_17)[0]
    scaled = dict(zip(SCALED_COLUMNS, scaled_17))

    row = {
        "WindSpeed_mps": scaled["WindSpeed_mps"], "WindDirection_deg": scaled["WindDirection_deg"],
        "AirTemperature_C": scaled["AirTemperature_C"], "Humidity_pct": scaled["Humidity_pct"],
        "AirPressure_hPa": scaled["AirPressure_hPa"], "GeneratorRPM": scaled["GeneratorRPM"],
        "GeneratorVoltage_V": scaled["GeneratorVoltage_V"], "GeneratorCurrent_A": scaled["GeneratorCurrent_A"],
        "BatteryVoltage_V": scaled["BatteryVoltage_V"], "PumpStatus": scaled["PumpStatus"],
        "PumpPower_W": scaled["PumpPower_W"], "WaterFlow_LPM": scaled["WaterFlow_LPM"],
        "TankLevel_pct": scaled["TankLevel_pct"], "EnergyGenerated_Wh": scaled["EnergyGenerated_Wh"],
        "EnergyConsumed_Wh": scaled["EnergyConsumed_Wh"],
        "Year": ts.year, "Month": ts.month, "Day": ts.day, "Hour": ts.hour,
        "WindPowerEfficiency": scaled["EnergyGenerated_Wh"] / (scaled["WindSpeed_mps"] + 1),
        "PumpEfficiency": scaled["WaterFlow_LPM"] / (scaled["PumpPower_W"] + 1),
        "EnergyEfficiency": scaled["EnergyGenerated_Wh"] / (scaled["EnergyConsumed_Wh"] + 1),
        "GeneratorLoadRatio": scaled["GeneratorCurrent_A"] / (scaled["GeneratorVoltage_V"] + 1),
        "WaterProductionEfficiency": scaled["WaterOutput_LPH"] / (scaled["PumpPower_W"] + 1),
    }
    X = pd.DataFrame([row])[MODEL_FEATURES]

    # --- Water output regression (ANN) ---
    scaled_pred = ann_reg_model.predict(X.values, verbose=0).flatten()[0]
    idx = SCALED_COLUMNS.index("WaterOutput_LPH")
    mean_, scale_ = scaler.mean_[idx], scaler.scale_[idx]
    water_output_lph = scaled_pred * scale_ + mean_

    # --- Fault detection ---
    fault_model = load_fault_model(fault_model_choice)
    if fault_model_choice == "RNN":
        X_input = X.values.reshape(1, 1, X.shape[1])
    elif fault_model_choice == "1D CNN":
        X_input = X.values.reshape(1, X.shape[1], 1)
    else:
        X_input = X.values
    fault_probs = fault_model.predict(X_input, verbose=0)[0]
    fault_class = int(np.argmax(fault_probs))
    fault_label = fault_code_encoder.inverse_transform([fault_class])[0]

    st.subheader("Results")
    c1, c2 = st.columns(2)
    c1.metric("Predicted Water Output", f"{water_output_lph:,.2f} LPH")
    c2.metric(f"Predicted Fault Code ({fault_model_choice})", str(fault_label),
              help=f"Confidence: {fault_probs[fault_class]*100:.1f}%")
    st.bar_chart(pd.Series(fault_probs, index=fault_code_encoder.classes_, name="Probability"))

st.sidebar.header("About")
st.sidebar.write(
    "ANN, RNN, and 1D CNN models were trained for both water-output regression and "
    "fault-code classification. This app uses the ANN model for regression (as in the "
    "notebook's Final Prediction section) and lets you pick which model to use for fault detection."
)
````
